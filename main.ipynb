##Ler o arquivo baixado

```python
#Descompacting the file
#Descompactando o arquivo
import pandas as pd

pasta_compactada = "C:\\Users\\Vitor\\Downloads\\Exportar-20260729T010332Z-1-001.zip"
pasta_destino="C:\\Users\\Vitor\\Downloads\\Descompactados"
caminho_arquivo="C:\\Users\\Vitor\\Downloads\\Descompactados\\Exportar\\Vendas - Dez.xlsx"

with zipfile.ZipFile(pasta_compactada,'r') as zip_ref:   # gerenciar recursos automaticamente. Ele garante que um recurso (arquivo, ZIP, conexão com banco de dados etc.) seja fechado ou liberado ao final do bloco, mesmo se ocorrer um erro.
    zip_ref.extractall(pasta_destino)
    
print("Arquivos descompactados com sucesso!")

df = pd.read_excel(caminho_arquivo)  

df.head() ## I could use display(df ) to show the entire table, but since it is very large, I used head() to show only the first 5 lines.

 ## eu poderia usar display(df) para mostrar a tabela inteira, mas como ela é muito grande, usei o head() para mostrar apenas as 5 primeiras linhas.
   
```

##Calcular indicadores (faturamento e quantidade de produtos vendidos)

```python
# Calculating the total revenue and quantity of products sold
# Calculando o faturamento total e a quantidade de produtos vendidos
faturamento=df['Valor Final'].sum()
qtde_produtos=df['Quantidade'].sum()

print(faturamento)
print(qtde_produtos)
```
